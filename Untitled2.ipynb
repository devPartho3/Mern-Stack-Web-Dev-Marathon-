```python
from prophet import Prophet
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/AirPassengers.csv')
df.head(5)
```

```python
df=df.rename(columns={'Month':'ds','Passengers':'y'})
df.head(5)
```

```python
df = df.rename(columns={'#Passengers': 'y'})
df['y_orig'] = df['y']
print(df.head(5))
```

```python
df['y']=np.log(df['y'])
df.head(5)
```

```python
model=Prophet()
model.fit(df)
```

```python
future_data=model.make_future_dataframe(periods=12,freq='M')
future_data.tail(12)
```

```python
forecast_data=model.predict(future_data)
forecast_data[['ds','yhat','yhat_lower','yhat_upper']].tail(12)
```

```python
model.plot_components(forecast_data)
```

```python
forecast_data_orig=forecast_data
forecast_data_orig['yhat']=np.exp(forecast_data_orig['yhat'])
forecast_data_orig['yhat_lower']=np.exp(forecast_data_orig['yhat_lower'])
forecast_data_orig['yhat_upper']=np.exp(forecast_data_orig['yhat_upper'])
forecast_data_orig
```

```python
model.plot(forecast_data_orig)
```

```python
df['y_log']=df['y']
df['y']=df['y_orig']

final_df=pd.DataFrame(forecast_data_orig[['ds','yhat','yhat_lower','yhat_upper']])
final_df.head(5)
```

```python
import plotly.graph_objects as go
import plotly.offline as py
```

```python
actual_chart=go.Scatter(x=df['ds'],y=df['y'],name='Actual')
forecast_chart=go.Scatter(x=final_df['ds'],y=final_df['yhat'],name='Forecast')
predict_chart=go.Scatter(x=final_df['ds'],y=final_df['yhat_lower'],name='Predict')
predict_chart_upper=go.Scatter(x=final_df['ds'],y=final_df['yhat_upper'],name='Predict_upper')
predict_chart_lower=go.Scatter(x=final_df['ds'],y=final_df['yhat_lower'],name='Predict_lower')

py.plot([actual_chart,forecast_chart,predict_chart,predict_chart_upper,predict_chart_lower])
#
```
